## Produtos

O notebook gera uma pasta `outputs/reports/integrated_v6/` contendo:

- `relatorio_integrado_2024.html` - versao visual para navegador;
- `relatorio_integrado_2024.tex` - fonte LaTeX completa;
- `relatorio_integrado_2024.pdf` - PDF compilado quando `latexmk` ou `pdflatex` estiver instalado;
- figuras do relatorio;
- tabelas de auditoria dos extremos.

Se LaTeX nao estiver instalado no Windows, o HTML continua sendo gerado normalmente e o `.tex` fica pronto para compilar depois.

```python
from pathlib import Path
import sys, subprocess

CWD = Path.cwd().resolve()
ROOT = CWD.parent if CWD.name.lower() == "notebooks" else CWD

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

script = ROOT / "utils" / "generate_integrated_report.py"
if not script.exists():
    raise FileNotFoundError(f"Nao encontrei {script}. Copie a pasta utils/ do pacote v6 para a raiz do projeto.")

cmd = [sys.executable, str(script), "--project-root", str(ROOT)]
print("Executando:", " ".join(cmd))
result = subprocess.run(cmd, text=True, capture_output=True)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
    raise RuntimeError("Falha ao gerar o relatorio integrado. Veja a mensagem acima.")
```

## Como abrir

Depois da execucao:

- abra `outputs/reports/integrated_v6/relatorio_integrado_2024.html` no navegador;
- se o PDF foi compilado, abra `relatorio_integrado_2024.pdf`;
- a fonte LaTeX fica ao lado para auditoria ou edicao manual.
