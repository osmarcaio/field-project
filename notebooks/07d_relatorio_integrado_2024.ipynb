# 07d - Relatórios integrados de desenvolvimento (2024)

Este notebook **não treina modelos**. Ele gera dois produtos complementares:

1. o relatório geral já existente, com comparação de modelos, extremos e incerteza;
2. um novo relatório v7 focado em **episódios, antecedência, persistência e falsos blocos de sinal**.

Execute `07c_avaliacao_operacional_por_episodio.ipynb` antes deste notebook.

```python
from pathlib import Path
import subprocess
import sys
import os

CWD = Path.cwd().resolve()
ROOT = CWD.parent if CWD.name.lower() == "notebooks" else CWD

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

script = ROOT / "utils" / "generate_integrated_report.py"

if not script.exists():
    raise FileNotFoundError(
        f"Não encontrei {script}. "
        "Confira se a pasta utils/ está na raiz do projeto."
    )

print("\n=== generate_integrated_report.py ===")
print(f"Script: {script}")
print()

env = os.environ.copy()

# Força o processo Python filho a usar UTF-8
env["PYTHONIOENCODING"] = "utf-8"
env["PYTHONUTF8"] = "1"

proc = subprocess.Popen(
    [
        sys.executable,
        "-u",
        str(script),
        "--project-root",
        str(ROOT),
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    encoding="utf-8",
    errors="replace",
    bufsize=1,
    env=env,
)

for line in proc.stdout:
    print(line, end="")

code = proc.wait()

if code != 0:
    raise RuntimeError(
        f"Falha em {script.name}; código {code}"
    )

print("\nRelatórios concluídos.")
```
